```python
import random
print("01_dicesimulator.")

def roll_dice():

    die1: int = random.randint(1, 6)
    die2: int = random.randint(1, 6)
    total: int = die1 + die2
    print(f"Total of two dice: {total}")

def die_Rolling():
    die1: int = 10
    print("die1 in die_Rolling starts as: " + str(die1))
    roll_dice()
    print("die1 in die_Rolling is: " + str(die1))

if __name__ == '__main__':
    die_Rolling()
```

```python
print("02_e=mc2")

C: int = 299792458

def energy():
    mass: int = float(input("Enter Mass in Kg: "))

    energy_joules: float = mass * (C ** 2)

    print("e = m * C^2")
    print("Mass = " + str(mass) + " kg")
    print("C = " + str(C)+ " m/s")

    print(f"{energy_joules} joules")


if __name__ == '__main__':
    energy()
```

```python
print("03_feet_to_inches")

def length_Convert():
    print("Feet to Inches Converter")
    feet = float(input("Enter feet: "))
    inches = feet * 12
    print(f"{feet} feet is {inches} inches!")

if __name__ == '__main__':
    length_Convert()
```

```python
print("04_pythagorean_theorem")
import math
def main():
    ab:float = float(input("Enter the Length of side AB: "))
    ac:float = float(input("Enter the Length of side AC: "))
    bc = math.sqrt(ab**2 + ac**2)
    print(f"The hypotenuse (BC) is: {bc:.2f} units")

if __name__ == '__main__':
    main()
```

```python
print("05_remainder_division")

def main():
    print("Division Calculator")
    num1:int = int(input("Enter the number to dividend: "))
    num2:int = int(input("Enter the number to divide by: "))
    quotient:int = num1 // num2
    remainder:int = num1 % num2
    print(f"Calculation is {num1} ÷ {num2} = {quotient} with a remainder of {remainder}")

if __name__ == '__main__':
    main()
```

```python
print("06_rolldice")

import random

def main():
    print("Rolling Two 6-Sided Dice")

    die1:int = random.randint(1, 6)
    die2:int = random.randint(1, 6)
    total:int = die1 + die2

    print(f"Die 1: {die1}")
    print(f"Die 2: {die2}")
    print(f"Total: {total}")

if __name__ == '__main__':
    main()
```

```python
print("06_seconds_in_year")

days_per_year: int = 365
hours_per_day: int = 24
min_per_day: int = 60
seconds_per_min: int = 60

def main():
    print("There are " + str(days_per_year * hours_per_day * min_per_day * seconds_per_min) + " seconds in a year!")

if __name__ == '__main__':
    main()
```
