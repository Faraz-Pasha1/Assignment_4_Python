```python
print("01_add_many_number")

def add_numbers(numbers)-> int:

    total_so_far: int = 0
    for number in numbers:
        total_so_far += number

    return total_so_far

def main():
    numbers: list[int] = [1, 2, 3, 4, 5]
    sum_of_numbers: int = add_numbers(numbers)
    print(sum_of_numbers)


if __name__ == '__main__':
    main()
```

```python
print("02_double_list")

def main():
    numbers: list[int] = [1, 2, 3, 4]

    for i in range(len(numbers)):
        elemurate = numbers[i]
        numbers[i] = elemurate * 2

    print(numbers)

if __name__ == '__main__':
    main()
```

```python
print("04_flowing_with_data_structures")

def add_copies(list, item):
    for i in range(3):
        list.append(item)

def main():
    message = input("Enter a message or word to make three copies: ")
    list = []
    print("List before:", list)
    add_copies(list, message)
    print("List after:", list)

if __name__ == "__main__":
    main()
```

```python
print("08_shorten")
max_length: int = 3

def shorten(lst):
    while len(lst) > max_length:
        last_elem = lst.pop()
        print("Removed:", last_elem)

def get_list():
    items = []
    elem = input("Please enter an element of the list or press enter to stop. ")
    while elem != "":
        items.append(elem)
        elem = input("Please enter an element of the list or press enter to stop. ")
    return items

def main():
    lst = get_list()
    shorten(lst)

if __name__ == '__main__':
    main()
```
