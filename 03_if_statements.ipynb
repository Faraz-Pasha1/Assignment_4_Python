```python
print("02_international_voting_age")

Pakistan_age : int = 16
China_age : int = 25
America_age : int = 48

def main():
    user_age = int(input("How old are you? "))
    if user_age >= Pakistan_age:
        print("You can vote in Pakistan where the voting age is " + str(Pakistan_age) + ".")
    else:
        print("You cannot vote in Pakistan where the voting age is " + str(Pakistan_age) + ".")

    if user_age >= China_age:
        print("You can vote in China where the voting age is " + str(China_age) + ".")
    else:
        print("You cannot vote in China where the voting age is " + str(China_age) + ".")

    if user_age >= America_age:
        print("You can vote in America where the voting age is " + str(America_age) + ".")
    else:
        print("You cannot vote in America where the voting age is " + str(America_age) + ".")

if __name__ == '__main__':
    main()
```

```python
print("03_leap_year")

def main():
    year = int(input('Please enter a year: '))

    if year % 4 == 0:
        if year % 100 == 0:
            if year % 400 == 0:
                print("That's a leap year!")
            else:
                print("That's not a leap year.")
        else:
            print("That's a leap year!")
    else:
        print("That's not a leap year.")

if __name__ == '__main__':
    main()
```

```python
print("04_tall_enough_to_ride")

min_Height: int = 50

def main():
    height = float(input("What is your Height? "))
    if height >= min_Height:
        print("Your height enough to ride!")
    else:
        print("Your heoght is not enough to ride, but maybe next year!")

if __name__ == '__main__':
    main()
```
