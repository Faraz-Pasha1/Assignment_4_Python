```python
print("01_add_two_numbers")
def main():
    print("Let's add two numbers!")
    num1 = int(input("Enter the first number: "))
    num2 = int(input("Enter the second number: "))
    total : int = num1 + num2
    print(f"The total is {total}.")

if __name__ == '__main__':
    main()
```

```python
print("03_fahrenheit_to_celsius")

def tempConvert():
  print("Converting °F to °C")
  degree_F = float(input("Enter Temperature in °F to convert it into °C = "))
  degree_C = (degree_F - 32) * 5.0/9.0
  print(f"Temperature of {degree_F} °F = {degree_C} °C")
if __name__ == '__main__':
    tempConvert()
```

```python
print("05_triangle_perimeter")

def triangle():
    print(" Triangle Perimeter Calculator")

    side1 = float(input("Enter length of 1st side of Trianle: "))
    side2 = float(input("Enter length of 2nd side of Trianle : "))
    side3 = float(input("Enter length of 3rd side of Trianle : "))

    total_Parameter = side1 + side2 + side3
    print(f"The total parameter of trianlge is: {total_Parameter} units")

if __name__ == '__main__':
    triangle()
```
